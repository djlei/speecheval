# 学生英语朗读自动评测（CrisperWhisper 逐字转写版 · modelscope）

本 notebook 用 **CrisperWhisper 2.0** 的逐字（verbatim）转写替换原来的 OpenAI Whisper。原版 Whisper 会把重复、口吃、"um/uh"这些不流利成分"顺"掉，CrisperWhisper 会照实写下来，例如：

`[um] he needs to to wear a bl- black uniform`

评测从三方面给出反馈，并按权重计入综合得分（默认 **准确度 70% + 流利度 20% + 识别置信度 10%**）：

- **准确度**：错读、漏读、多读。重复和自我修正**不再算作"多读"**，不扣准确度。
- **流利度**：语速；**句中停顿**和**句间停顿**分开统计，只有句中停顿和过长的停顿扣分；以及**重复、自我修正、断词、填充词**的次数和位置。
- **识别置信度**：模型对学生读对的词有多大把握，可以粗略反映发音是否清晰。

**使用步骤**

1. 菜单 **代码执行程序 / Runtime → 更改运行时类型 / Change runtime type → T4 GPU**。CPU 也能跑，只是慢很多。
2. 在「1. 参数配置」里选择数据来源：直接从 GitHub 仓库拉取（默认）、Google Drive，或手动上传。
3. **代码执行程序 / Runtime → 全部运行 / Run all**。报告输出到 `output/` 目录，最后会自动打包下载。

**数据约定**（与原仓库一致）：朗读原文放在 `original.txt`，学生录音放在 `audio/` 目录，**文件名即学生姓名**，支持 mp3 / m4a / wav / flac / ogg / aac / mp4 / webm。

## 3. 工具函数

```python
import re, json, csv, string, difflib
from collections import Counter
import numpy as np

PUNCT = string.punctuation + "’‘“”…—–"
_PUNCT_TABLE = str.maketrans("", "", PUNCT)
FILLERS = {"um", "umm", "uh", "uhm", "er", "erm", "ah", "eh", "hmm", "mm", "mhm"}


def normalize_word(w):
    # 小写、去标点（含撇号，o'clock → oclock，原文和识别结果处理方式一致）
    return w.lower().translate(_PUNCT_TABLE).strip()


BOUNDARY_PUNCT = ".,!?;:…—–"


def preprocess_text_with_boundaries(text):
    # 原文标准化（小写、去标点、连字符拆开），同时记录每个词后面是否是句读处：
    # 词后带 . , ! ? ; : 等标点，或者是段落（空行分隔）的最后一个词。
    # 注意：不使用普通换行判断，因为绘本排版的换行常常落在短语中间。
    words, boundary = [], []
    for para in re.split(r"\n\s*\n", text):
        n0 = len(words)
        for tok in para.split():
            core = tok.strip("\"“”'‘’()[]")
            ends = bool(core) and core[-1] in BOUNDARY_PUNCT
            parts = [n for n in (normalize_word(p) for p in re.split(r"[-–—]", tok)) if n]
            if not parts:
                if ends and words:
                    boundary[-1] = True
                continue
            for k, p in enumerate(parts):
                words.append(p)
                boundary.append(ends and k == len(parts) - 1)
        if len(words) > n0:
            boundary[-1] = True
    # 第一行没有结尾标点时视为标题（如 "People who help us"），标题末尾也算一处句读
    lines = [l for l in text.splitlines() if l.strip()]
    if lines and len(lines) > 1 and lines[0].rstrip()[-1] not in BOUNDARY_PUNCT:
        n_title = len(preprocess_text_with_boundaries(lines[0])[0])
        if 0 < n_title <= len(boundary):
            boundary[n_title - 1] = True
    return words, boundary


def preprocess_text(text):
    return preprocess_text_with_boundaries(text)[0]


def fmt_time(sec):
    if sec is None:
        return "--:--"
    return f"{int(sec // 60):02d}:{sec % 60:04.1f}"


# ---------------------------------------------------------------
# 音频解码：用 ffmpeg 转成 16 kHz 单声道（soundfile 不支持 m4a/mp3 等格式）
# ---------------------------------------------------------------
def load_audio_16k(path):
    cmd = ["ffmpeg", "-nostdin", "-v", "error", "-i", path,
           "-f", "f32le", "-ac", "1", "-ar", "16000", "-"]
    out = subprocess.run(cmd, capture_output=True, check=True).stdout
    return np.frombuffer(out, np.float32).copy()


# ---------------------------------------------------------------
# 把 CrisperWhisper 输出的每个词归类：
#   word   普通单词        filler  填充词 [um] [uh]
#   cutoff 断词 "th-"      sound   其他声音 [laughter] [cough] ...
# ---------------------------------------------------------------
def classify_tokens(raw_words, ref_vocab):
    tokens = []
    for w in raw_words:
        raw = (w.get("word") or "").strip()
        start, end, prob = w.get("start"), w.get("end"), w.get("prob")
        t = raw.strip(" .,!?;:\"“”'‘’").lower()
        if not t:
            continue
        if t.startswith("[") and t.endswith("]"):
            inner = t[1:-1].strip()
            kind = "filler" if inner in FILLERS else "sound"
            tokens.append(dict(kind=kind, text=f"[{inner}]", norm=inner, start=start, end=end, prob=prob))
            continue
        if len(t) > 1 and t.endswith("-"):
            tokens.append(dict(kind="cutoff", text=t, norm=normalize_word(t), start=start, end=end, prob=prob))
            continue
        # 普通词：内部连字符拆开（如 well-known），时间平均分配
        parts = [p for p in (normalize_word(x) for x in re.split(r"[-\s]+", t)) if p]
        if not parts:
            continue
        span = (end - start) / len(parts) if (start is not None and end is not None) else None
        for k, p in enumerate(parts):
            s = start + k * span if span is not None else start
            e = s + span if span is not None else end
            kind = "filler" if (p in FILLERS and p not in ref_vocab) else "word"
            tokens.append(dict(kind=kind, text=p, norm=p, start=s, end=e, prob=prob))
    return tokens


# ---------------------------------------------------------------
# 词级最小编辑距离对齐（比 difflib 更适合做朗读比对）
# 返回 [(tag, ref_idx, hyp_idx)]，tag ∈ equal / replace / insert / delete
# insert 的 ref_idx 表示"插在原文第 ref_idx 个词之前"
# ---------------------------------------------------------------
def align_words(ref, hyp):
    n, m = len(ref), len(hyp)
    D = [[0] * (m + 1) for _ in range(n + 1)]
    for i in range(1, n + 1):
        D[i][0] = i
    for j in range(1, m + 1):
        D[0][j] = j
    for i in range(1, n + 1):
        ri = ref[i - 1]
        for j in range(1, m + 1):
            c = 0 if ri == hyp[j - 1] else 1
            D[i][j] = min(D[i - 1][j - 1] + c, D[i - 1][j] + 1, D[i][j - 1] + 1)
    i, j, pairs = n, m, []
    while i > 0 or j > 0:
        if i > 0 and j > 0 and D[i][j] == D[i - 1][j - 1] + (0 if ref[i - 1] == hyp[j - 1] else 1):
            pairs.append(("equal" if ref[i - 1] == hyp[j - 1] else "replace", i - 1, j - 1))
            i, j = i - 1, j - 1
        elif j > 0 and D[i][j] == D[i][j - 1] + 1:
            pairs.append(("insert", i, j - 1))
            j -= 1
        else:
            pairs.append(("delete", i - 1, None))
            i -= 1
    return pairs[::-1]


def group_segments(pairs):
    # 把连续的同类操作合并成段：dict(tag, ref=[...], hyp=[...], pos)
    segs = []
    for tag, ri, hj in pairs:
        last = segs[-1] if segs else None
        if last and last["tag"] == tag and tag != "equal" and (tag != "insert" or last["pos"] == ri):
            if ri is not None and tag != "insert":
                last["ref"].append(ri)
            if hj is not None:
                last["hyp"].append(hj)
        else:
            segs.append(dict(tag=tag, pos=ri,
                             ref=[] if (tag == "insert" or ri is None) else [ri],
                             hyp=[] if hj is None else [hj]))
    return segs


def _contains(seq, sub):
    k = len(sub)
    return any(seq[i:i + k] == sub for i in range(len(seq) - k + 1))


def classify_insert(seg, ref, hyp):
    # 判断一段"多出来的词"是重复、自我修正还是真正的多读
    S = [hyp[j] for j in seg["hyp"]]
    k, p = len(S), seg["pos"]
    j1, j2 = seg["hyp"][0], seg["hyp"][-1] + 1
    # 1) 与紧邻的实际读出内容相同 → 重复（如 to to、he needs to he needs to）
    if hyp[max(0, j1 - k):j1] == S or hyp[j2:j2 + k] == S:
        return "repeat"
    # 2) 是附近原文的一段 → 回读/重复
    if _contains(ref[max(0, p - k - 2):p + k + 2], S):
        return "repeat"
    # 3) 单个词且与相邻原文词很像（如 need → needs）→ 自我修正
    if k == 1:
        for r in ref[max(0, p - 1):p + 1]:
            if (len(S[0]) >= 2 and (r.startswith(S[0]) or S[0].startswith(r))) \
                    or difflib.SequenceMatcher(None, S[0], r).ratio() >= 0.7:
                return "selfcorr"
    return "extra"
```

```python
# ---------------------------------------------------------------
# 核心评测：对齐 + 准确度 + 流利度指标 + 标注文本
# ---------------------------------------------------------------
def evaluate(raw_words, ref_words, ref_boundary):
    ref_vocab = set(ref_words)
    tokens = classify_tokens(raw_words, ref_vocab)
    core_idx = [i for i, t in enumerate(tokens) if t["kind"] == "word"]   # 参与对齐的词
    hyp = [tokens[i]["norm"] for i in core_idx]

    pairs = align_words(ref_words, hyp)
    segs = group_segments(pairs)

    labels = {}            # 对齐用的 hyp 下标 → equal/replace/repeat/selfcorr/extra
    errors, repeats, selfcorrs = [], [], []
    last_t = None          # 上一个已读词的结束时间，用来给漏读标注大致位置
    for s in segs:
        t0 = tokens[core_idx[s["hyp"][0]]]["start"] if s["hyp"] else last_t
        if s["hyp"]:
            last_t = tokens[core_idx[s["hyp"][-1]]]["end"]
        if s["tag"] == "equal":
            for j in s["hyp"]:
                labels[j] = "equal"
        elif s["tag"] == "replace":
            for j in s["hyp"]:
                labels[j] = "replace"
            errors.append(("错读", t0, f"原文「{' '.join(ref_words[i] for i in s['ref'])}」"
                                        f" → 读成「{' '.join(hyp[j] for j in s['hyp'])}」"))
        elif s["tag"] == "delete":
            errors.append(("漏读", t0, f"原文「{' '.join(ref_words[i] for i in s['ref'])}」未朗读"))
        else:  # insert
            kind = classify_insert(s, ref_words, hyp)
            for j in s["hyp"]:
                labels[j] = kind
            text = " ".join(hyp[j] for j in s["hyp"])
            if kind == "repeat":
                repeats.append((t0, text))
            elif kind == "selfcorr":
                selfcorrs.append((t0, text))
            else:
                errors.append(("多读", t0, f"额外读出「{text}」"))

    correct = sum(1 for tg, _, _ in pairs if tg == "equal")
    accuracy = correct / len(ref_words) * 100 if ref_words else 0.0

    # ---------- 停顿：两个单词之间的"犹豫"，区分句中 / 句间 ----------
    core_pos = {ti: j for j, ti in enumerate(core_idx)}
    tag_of = {hj: tg for tg, ri, hj in pairs if hj is not None}
    ref_of = {hj: ri for tg, ri, hj in pairs if hj is not None}   # insert 为插入点

    def left_ref(j):   # 这个词"之后"对应原文的位置
        return ref_of[j] - 1 if tag_of[j] == "insert" else ref_of[j]

    hesitations, after = [], {}
    for a, b in zip(core_idx, core_idx[1:]):
        A, B = tokens[a], tokens[b]
        if A["end"] is None or B["start"] is None:
            continue
        dur = B["start"] - A["end"]
        if dur <= PAUSE_THRESHOLD:
            continue
        lo, hi = max(left_ref(core_pos[a]), 0), ref_of[core_pos[b]]
        is_boundary = any(ref_boundary[k] for k in range(lo, min(hi, len(ref_boundary))))
        h = dict(start=A["end"], dur=dur, boundary=is_boundary, prev=A["text"], next=B["text"],
                 between=[tokens[x]["text"] for x in range(a + 1, b)])
        h["long"] = dur > (LONG_BOUNDARY_PAUSE if is_boundary else LONG_HESITATION)
        hesitations.append(h)
        after[a] = h

    # ---------- 标注文本（按实际朗读顺序，漏读插在对应位置） ----------
    out, emitted = [], 0
    word_fmt = {"equal": "{w}", "replace": "【{w}】", "repeat": "⟨重复:{w}⟩",
                "selfcorr": "⟨修正:{w}⟩", "extra": "{{多读:{w}}}"}

    def emit_token(ti, text=None):
        t = tokens[ti]
        if text is not None:
            out.append(text)
        elif t["kind"] == "word":
            out.append(word_fmt[labels.get(core_pos[ti], "equal")].format(w=t["text"]))
        elif t["kind"] == "cutoff":
            out.append(f"⟨断词:{t['text']}⟩")
        else:
            out.append(t["text"])
        h = after.get(ti)               # 这个词之后的停顿标记
        if h:
            if not h["boundary"]:
                out.append(f"‖{h['dur']:.1f}s‖" if h["long"] else "/")
            elif h["long"]:
                out.append(f"‖句间{h['dur']:.1f}s‖")

    for tg, ri, hj in pairs:
        if hj is None:                       # 漏读
            out.append(f"（漏:{ref_words[ri]}）")
            continue
        ti = core_idx[hj]
        while emitted < ti:                  # 先输出夹在中间的填充词/断词/声音
            if tokens[emitted]["kind"] != "word":
                emit_token(emitted)
            emitted += 1
        emit_token(ti, f"【{ref_words[ri]}→{tokens[ti]['text']}】" if tg == "replace" else None)
        emitted = ti + 1
    while emitted < len(tokens):
        if tokens[emitted]["kind"] != "word":
            emit_token(emitted)
        emitted += 1
    annotated = " ".join(out)
    # 合并连续的重复标注：⟨重复:who⟩ ⟨重复:help⟩ → ⟨重复:who help⟩
    while True:
        merged = re.sub(r"⟨重复:([^⟩]+)⟩ ⟨重复:([^⟩]+)⟩", r"⟨重复:\1 \2⟩", annotated)
        if merged == annotated:
            break
        annotated = merged

    # ---------- 流利度指标 ----------
    timed = [t for t in tokens if t["start"] is not None and t["end"] is not None]
    total_duration = (timed[-1]["end"] - timed[0]["start"]) if timed else 0.0
    n_fillers = sum(1 for t in tokens if t["kind"] == "filler")
    n_cutoffs = sum(1 for t in tokens if t["kind"] == "cutoff")
    effective_words = sum(1 for lab in labels.values() if lab in ("equal", "replace"))
    n_disfl = len(repeats) + len(selfcorrs) + n_cutoffs + n_fillers
    minutes = total_duration / 60 if total_duration > 0 else 0
    per100 = 100 / len(ref_words) if ref_words else 0

    within = [h for h in hesitations if not h["boundary"]]
    between = [h for h in hesitations if h["boundary"]]
    long_within = [h for h in within if h["long"]]
    long_between = [h for h in between if h["long"]]
    burden = (sum(min(h["dur"], HESITATION_CAP) for h in long_within)
              + sum(min(h["dur"] - LONG_BOUNDARY_PAUSE, HESITATION_CAP) for h in long_between)) * per100
    avg = lambda hs: round(sum(h["dur"] for h in hs) / len(hs), 2) if hs else 0
    boundary_slots = sum(ref_boundary[:-1])

    pause = dict(n_within=len(within), within_per100=round(len(within) * per100, 1),
                 n_between=len(between), boundary_slots=boundary_slots,
                 long_within=long_within, long_between=long_between, burden=round(burden, 2))

    fluency_data = {
        "总朗读时长(秒)": round(total_duration, 2),
        "语速(WPM，不含重复/填充词)": round(effective_words / minutes, 1) if minutes else 0,
        "每分钟正确词数(WCPM)": round(correct / minutes, 1) if minutes else 0,
        "停顿总次数(>%.2fs)" % PAUSE_THRESHOLD: len(hesitations),
        "句间停顿次数（原文句读处，不扣分）": f"{len(between)}（原文共 {boundary_slots} 处句读）",
        "句间停顿平均时长(秒)": avg(between),
        "句中停顿次数": len(within),
        "句中停顿平均时长(秒)": avg(within),
        "每百词句中停顿次数": pause["within_per100"],
        "句中长犹豫次数(>%.1fs)" % LONG_HESITATION: len(long_within),
        "句间过长停顿次数(>%.1fs)" % LONG_BOUNDARY_PAUSE: len(long_between),
        "长停顿负担(秒/百词)": pause["burden"],
        "重复次数": len(repeats),
        "自我修正次数": len(selfcorrs),
        "断词次数": n_cutoffs,
        "填充词次数(um/uh等)": n_fillers,
        "每百词不流利次数": round(n_disfl * per100, 1),
    }

    # ---------- 识别置信度：只看读对的词（错读已经在准确度里扣过分） ----------
    correct_tokens = [tokens[ti] for j, ti in enumerate(core_idx)
                      if labels.get(j) == "equal" and tokens[ti]["prob"] is not None]
    low_conf = [(t["start"], t["text"], t["prob"]) for t in correct_tokens
                if t["prob"] < LOW_CONF_THRESHOLD]
    if correct_tokens:
        conf_data = {
            "参与统计的正确词数": len(correct_tokens),
            "平均识别置信度": round(sum(t["prob"] for t in correct_tokens) / len(correct_tokens), 3),
            "低置信度词数(<%.2f)" % LOW_CONF_THRESHOLD: len(low_conf),
            "低置信度词占比(%)": round(len(low_conf) / len(correct_tokens) * 100, 1),
        }
    else:
        conf_data = None   # 没有置信度数据（例如旧缓存没有 prob 字段）

    return dict(tokens=tokens, accuracy=accuracy, correct=correct, errors=errors,
                repeats=repeats, selfcorrs=selfcorrs, pause=pause,
                annotated=annotated, fluency_data=fluency_data,
                low_conf=low_conf, conf_data=conf_data)
```

```python
def linear_score(x, good, bad, pts):
    # 连续计分：x ≤ good 满分，x ≥ bad 零分，中间线性插值（避免分档"断崖"）
    if x <= good:
        return float(pts)
    if x >= bad:
        return 0.0
    return pts * (bad - x) / (bad - good)


def fluency_grade(ev):
    # 流利度评分（满分 100）：语速 25 + 句中停顿 25 + 长停顿 20 + 重复与不流利 30
    fd, pz, repeats = ev["fluency_data"], ev["pause"], ev["repeats"]
    wpm = fd["语速(WPM，不含重复/填充词)"]
    disfl = fd["每百词不流利次数"]
    fb, score = [], 0.0

    if wpm >= 100:   fb.append("✅ 语速较快，朗读节奏流畅"); score += 25
    elif wpm >= 70:  fb.append("✅ 语速适中，节奏平稳自然"); score += 22
    elif wpm >= 50:  fb.append("⚠️ 语速偏慢，存在较多思考间隙"); score += 15
    else:            fb.append("❌ 语速过慢，连贯性不足"); score += 8

    # 句中停顿（25 分）
    s_within = linear_score(pz["within_per100"], WITHIN_PAUSE_GOOD, WITHIN_PAUSE_BAD, 25)
    rate = f"共 {pz['n_within']} 处，每百词 {pz['within_per100']} 次"
    if s_within >= 21:   fb.append(f"✅ 句中停顿很少（{rate}），能按意群连贯朗读")
    elif s_within >= 12: fb.append(f"⚠️ 句中停顿偏多（{rate}），部分短语读得不够连贯，标注中“/”处即停顿位置")
    else:                fb.append(f"❌ 句中停顿频繁（{rate}），朗读断断续续，建议按意群练习")
    score += s_within

    # 长停顿（20 分）：句中长犹豫 + 句间过长停顿，按时长加权
    s_long = linear_score(pz["burden"], LONG_BURDEN_GOOD, LONG_BURDEN_BAD, 20)
    n_lw, n_lb = len(pz["long_within"]), len(pz["long_between"])
    if n_lw + n_lb == 0:
        fb.append("✅ 没有明显的长时间卡顿")
    else:
        msg = f"句中长犹豫 {n_lw} 处、句间过长停顿 {n_lb} 处（长停顿负担 {pz['burden']} 秒/百词），位置见上方列表"
        fb.append(("⚠️ " if s_long >= 14 else "❌ ") + msg)
    score += s_long

    # 句间停顿：不计分，只给朗读节奏上的提示
    if pz["boundary_slots"] >= 5 and pz["n_between"] < 0.3 * pz["boundary_slots"]:
        fb.append(f"💡 原文 {pz['boundary_slots']} 处句读中只在 {pz['n_between']} 处有停顿，"
                  "可以在句号、逗号处稍作停顿，读出句子的节奏（本项不扣分）")

    detail = (f"重复{fd['重复次数']}次、自我修正{fd['自我修正次数']}次、"
              f"断词{fd['断词次数']}次、填充词{fd['填充词次数(um/uh等)']}次")
    if disfl <= 2:    fb.append(f"✅ 几乎没有重复和卡顿（{detail}）"); score += 30
    elif disfl <= 5:  fb.append(f"⚠️ 有少量重复或卡顿（{detail}）"); score += 22
    elif disfl <= 10: fb.append(f"❌ 重复和卡顿较多（{detail}）"); score += 14
    else:             fb.append(f"❌ 重复和卡顿频繁（{detail}），建议先熟读课文"); score += 6

    common = Counter(text for _, text in repeats).most_common(3)
    if common and common[0][1] >= 2:
        fb.append("💡 反复重复的地方：" + "、".join(f"「{w}」×{c}" for w, c in common if c >= 2))

    score = round(score, 1)
    level = "优秀" if score >= 85 else "良好" if score >= 70 else "一般" if score >= 50 else "待加强"
    return {"流利度得分": score, "流利度等级": level, "详细评价": fb}


def confidence_grade(conf_data):
    # 识别置信度得分（满分 100）= 读对的词的平均识别置信度 × 100
    if not conf_data:
        return None
    score = round(conf_data["平均识别置信度"] * 100, 1)
    ratio = conf_data["低置信度词占比(%)"]
    if score >= 90:   level, fb = "优秀", "✅ 发音清晰，模型对绝大多数单词都很有把握"
    elif score >= 80: level, fb = "良好", "✅ 发音基本清晰，个别单词可以读得更饱满"
    elif score >= 70: level, fb = "一般", "⚠️ 部分单词发音不够清晰，建议对照下方列出的词回听练习"
    else:             level, fb = "待加强", "❌ 较多单词发音含糊，建议放慢语速、把每个音读完整"
    return {"识别置信度得分": score, "识别置信度等级": level,
            "详细评价": [fb, f"低置信度词占读对词的 {ratio}%"]}


def combine_scores(accuracy, fluency_score, conf_score):
    # 加权综合得分；没有置信度数据时，按准确度:流利度原比例重新分配权重
    if conf_score is None:
        w = ACCURACY_WEIGHT + FLUENCY_WEIGHT
        return accuracy * ACCURACY_WEIGHT / w + fluency_score * FLUENCY_WEIGHT / w
    return accuracy * ACCURACY_WEIGHT + fluency_score * FLUENCY_WEIGHT + conf_score * CONFIDENCE_WEIGHT


def total_comment(total):
    if total >= 90: return "整体表现优秀，发音准确且朗读流畅，继续保持。"
    if total >= 75: return "整体表现良好，针对少量错误和停顿加强练习即可明显提升。"
    if total >= 60: return "整体表现一般，建议先纠正易错单词，再逐步提升朗读流畅度。"
    return "整体有待加强，建议从单词发音基础开始练习，熟练后再提升语速。"


def generate_report(name, verbatim_text, ev, ref_words, fl, cf, total):
    L = []
    L += ["=" * 60, "学生朗读评估报告", f"学生姓名：{name}", f"转写模型：CrisperWhisper 2.0 ({MODEL_SIZE}, verbatim)", "=" * 60, ""]
    L += ["【一、逐字转写（保留重复、填充词、断词）】", verbatim_text.strip(), ""]
    L += ["【二、朗读标注】",
          "  图例：【原词→读成】错读  （漏:词）漏读  {多读:词}  ⟨重复:词⟩  ⟨修正:词⟩  ⟨断词:th-⟩  [um]填充词",
          "        /句中停顿  ‖1.5s‖句中长犹豫  ‖句间2.6s‖句间过长停顿（句号、逗号处的正常停顿不标出）",
          ev["annotated"], ""]
    L += ["【三、发音准确度评估】",
          f"原文总单词数：{len(ref_words)}",
          f"正确朗读单词数：{ev['correct']}",
          f"准确率：{ev['accuracy']:.1f}%（重复和自我修正不扣准确度分）", "", "错误明细："]
    if ev["errors"]:
        for kind, t, msg in ev["errors"]:
            icon = {"错读": "❌", "漏读": "⚠️", "多读": "➕"}[kind]
            L.append(f"  {icon} {kind}{'' if t is None else f' [{fmt_time(t)}]'}：{msg}")
    else:
        L.append("  ✅ 全部匹配，未发现错读、漏读、多读")
    L.append("")
    L.append("【四、流利度指标详情】")
    L += [f"  {k}：{v}" for k, v in ev["fluency_data"].items()]
    if ev["repeats"] or ev["selfcorrs"]:
        L.append("  重复 / 自我修正位置：")
        items = [(t, "重复", w) for t, w in ev["repeats"]] + [(t, "修正", w) for t, w in ev["selfcorrs"]]
        for t, kind, w in sorted(items, key=lambda x: x[0] if x[0] is not None else 0):
            L.append(f"    [{fmt_time(t)}] {kind}「{w}」")
    for title, hs in (("句中长犹豫位置：", ev["pause"]["long_within"]),
                      ("句间过长停顿位置：", ev["pause"]["long_between"])):
        if hs:
            L.append("  " + title)
            for h in hs:
                mid = f"（中间有 {' '.join(h['between'])}）" if h["between"] else ""
                L.append(f"    [{fmt_time(h['start'])}] 「{h['prev']}」与「{h['next']}」之间{mid}停顿 {h['dur']:.1f} 秒")
    L.append("")
    L += ["【五、流利度评价】",
          f"  流利度得分：{fl['流利度得分']}/100", f"  流利度等级：{fl['流利度等级']}", "  详细说明："]
    L += [f"    {x}" for x in fl["详细评价"]]
    L.append("")
    L.append("【六、识别置信度评估（反映发音清晰程度，仅供参考）】")
    if cf:
        L += [f"  {k}：{v}" for k, v in ev["conf_data"].items()]
        L += [f"  识别置信度得分：{cf['识别置信度得分']}/100",
              f"  识别置信度等级：{cf['识别置信度等级']}", "  详细说明："]
        L += [f"    {x}" for x in cf["详细评价"]]
        if ev["low_conf"]:
            L.append("  置信度较低的词（建议回听确认）：")
            L.append("    " + "、".join(f"{w}[{fmt_time(t)}]"
                                        for t, w, _ in sorted(ev["low_conf"], key=lambda x: x[2])[:15]))
    else:
        L.append("  ⚠️ 没有置信度数据（可能是旧的缓存文件），本项不计分，权重按比例分给准确度和流利度")
    L.append("")
    L.append("【七、综合总评】")
    if cf:
        L += [f"  综合得分：{total:.1f}/100",
              f"    = 准确度 {ev['accuracy']:.1f} × {ACCURACY_WEIGHT:.0%}"
              f" + 流利度 {fl['流利度得分']} × {FLUENCY_WEIGHT:.0%}"
              f" + 识别置信度 {cf['识别置信度得分']} × {CONFIDENCE_WEIGHT:.0%}"]
    else:
        w = ACCURACY_WEIGHT + FLUENCY_WEIGHT
        L += [f"  综合得分：{total:.1f}/100",
              f"    = 准确度 {ev['accuracy']:.1f} × {ACCURACY_WEIGHT / w:.0%}"
              f" + 流利度 {fl['流利度得分']} × {FLUENCY_WEIGHT / w:.0%}（无置信度数据）"]
    L += [f"  {total_comment(total)}", "", "=" * 60]
    return "\n".join(L) + "\n"
```
